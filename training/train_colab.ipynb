## Train
Synthetic only: ~10 min on a T4. With a real dataset, raise epochs to 30–40.

```
args = '--synthetic 30000 --epochs 25 --bs 64'
if REAL: args += f' --real "{REAL}"'
if REAL_VAL: args += f' --real-val "{REAL_VAL}"'
!python train.py {args}
```
